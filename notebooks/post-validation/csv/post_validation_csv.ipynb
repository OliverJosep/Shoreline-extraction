### Analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('shoreline_validation_results_expermient1-3.csv')

# Load from experiment 3 only
exp3_filter = df['experiment'].str.contains('exp3', case=False)
dir_filter = df['direction'].isin(['rectified-truth', 'oblique-truth'])
df_exp3 = df[exp3_filter & dir_filter].copy()

# clean direction names
df_exp3['Resolution'] = df_exp3['experiment'].str.replace('exp3_deeplabv3_', '')

site_mapping = {
    'agrelo': 'Agrelo',
    'arenaldentem': "Arenal d'en Tem",
    'cadiz': 'Cádiz',
    'cies': 'Cíes',
    'samarador': "S'Amarador",
    'all': 'All Sites'
}
df_exp3['Site_Label'] = df_exp3['site'].map(site_mapping).fillna(df_exp3['site'].str.title())

metrics = ['MAE', 'RMSE', 'Q3', 'SD']
df_melted = df_exp3.melt(
    id_vars=['Site_Label', 'direction', 'Resolution'], 
    value_vars=metrics, 
    var_name='Metric', 
    value_name='Value (m)'
)

sns.set_style("whitegrid")

g = sns.catplot(
    data=df_melted, 
    kind="bar",
    x="Site_Label", 
    y="Value (m)", 
    hue="Resolution",
    col="Metric", 
    row="direction",
    height=3.5, 
    aspect=1.2,
    palette="viridis", # O 'mako', 'rocket', etc.
    sharey=False,
    margin_titles=True
)

g.set_titles(row_template="{row_name}", col_template="{col_name}")
g.set_axis_labels("", "Error (m)")
for ax in g.axes.flat:
    ax.tick_params(axis='x', rotation=45) 

plt.subplots_adjust(top=0.9)
g.figure.suptitle('Experiment 3 Analysis: Metrics per Site, Direction & Resolution', fontsize=16)

plt.show()
```

### Calculate Improvements 

```python
import pandas as pd

def calculate_improvements(experiments_csv, baseline_csv):
    # 1. Load DataFrames
    df_exp = pd.read_csv(experiments_csv)
    df_base = pd.read_csv(baseline_csv)
    
    # 2. Extract the SINGLE Reference Value: Baseline (Rectified-Truth)
    # We only want the 'rectified-truth' MAE from the Baseline experiment.
    baseline_ref_row = df_base[
        (df_base['site'] == 'all') & 
        (df_base['direction'] == 'rectified-truth')
    ]
    
    # Safety check
    if baseline_ref_row.empty:
        print(f"ERROR: Could not find 'rectified-truth' in baseline file.")
        return

    # Get the scalar value (e.g., 15.4)
    mae_baseline_val = baseline_ref_row['MAE'].values[0]
    
    print(f"INFO: Global Baseline Reference (Rectified) = {mae_baseline_val}")

    # 3. Define target directions for the Models
    target_directions = ['rectified-truth', 'oblique-truth']
    
    # Filter Experiments Data (Global site + Target directions)
    df_exp_filtered = df_exp[
        (df_exp['site'] == 'all') & 
        (df_exp['direction'].isin(target_directions))
    ].copy()

    # Columns to display
    cols = ['experiment', 'direction', 'MAE_baseline_rect', 'MAE_model', 'Imp (%)']

    # =========================================================
    # COMPARISON 1: EXP 1 (BiLSTM) vs BASELINE (RECTIFIED)
    # =========================================================
    print("-" * 80)
    print(f"IMPROVEMENT: EXP 1 (BiLSTM) vs BASELINE (Rectified-Truth)")
    print("-" * 80)
    
    # Filter for BiLSTM
    exp1_data = df_exp_filtered[df_exp_filtered['experiment'] == 'exp1_bilstm'].copy()
    
    # Assign the fixed Baseline Rectified value to all rows
    exp1_data['MAE_baseline_rect'] = mae_baseline_val
    exp1_data.rename(columns={'MAE': 'MAE_model'}, inplace=True)
    
    # Calculate Improvement Percentage
    exp1_data['Imp (%)'] = (
        (exp1_data['MAE_baseline_rect'] - exp1_data['MAE_model']) / exp1_data['MAE_baseline_rect']
    ) * 100
    
    # Print Results
    print(exp1_data[cols].sort_values(['direction']).round(1).to_string(index=False))
    print("\n")

    # =========================================================
    # COMPARISON 2: EXP 2 (CNNs) vs BASELINE (RECTIFIED)
    # =========================================================
    print("-" * 80)
    print(f"IMPROVEMENT: EXP 2 (CNNs) vs BASELINE (Rectified-Truth)")
    print("-" * 80)
    
    # Filter for Experiment 2 models
    exp2_data = df_exp_filtered[df_exp_filtered['experiment'].str.contains('exp2', na=False)].copy()
    
    # Assign the fixed Baseline Rectified value to all rows
    exp2_data['MAE_baseline_rect'] = mae_baseline_val
    exp2_data.rename(columns={'MAE': 'MAE_model'}, inplace=True)
    
    # Calculate Improvement Percentage
    exp2_data['Imp (%)'] = (
        (exp2_data['MAE_baseline_rect'] - exp2_data['MAE_model']) / exp2_data['MAE_baseline_rect']
    ) * 100
    
    # Print Results
    print(exp2_data[cols].sort_values(['experiment', 'direction']).round(1).to_string(index=False))

# --- EXECUTION ---
calculate_improvements(
    experiments_csv='shoreline_validation_results_expermient1-3.csv', 
    baseline_csv='shoreline_validation_results_baseline.csv'
)
```

```python
import pandas as pd

# script to calculate and display improvements between experiments
def calculate_improvements(csv_path):
    df = pd.read_csv(csv_path)
    
    df_global = df[df['site'] == 'all'].copy()
    
    target_directions = ['rectified-truth', 'oblique-truth']
    df_global = df_global[df_global['direction'].isin(target_directions)]

    # # Print results
    cols = ['experiment', 'direction', 'MAE_baseline', 'MAE_model', 'Imp (%)']

    print("-" * 60)
    print("IMPROVEMENT: EXP 3 (Resolutions) vs 256x256")
    print("-" * 60)
    
    # Baseline for Experiment 3
    baseline_name_exp3 = 'exp3_deeplabv3_256x256' 
    baseline_exp3 = df_global[df_global['experiment'] == baseline_name_exp3]
    
    if baseline_exp3.empty:
        print(f"Baseline experiment '{baseline_name_exp3}' not found in the data.")
    else:
        # Define other models for Experiment 3
        exp3_models = df_global[
            (df_global['experiment'].str.contains('exp3', na=False)) & 
            (df_global['experiment'] != baseline_name_exp3)
        ]
        
        merged_exp3 = pd.merge(
            exp3_models, 
            baseline_exp3[['direction', 'MAE']], 
            on='direction', 
            suffixes=('_model', '_baseline')
        )
        
        # Compute % Improvement
        merged_exp3['Imp (%)'] = (
            (merged_exp3['MAE_baseline'] - merged_exp3['MAE_model']) / merged_exp3['MAE_baseline']
        ) * 100
        
        print(merged_exp3[cols].sort_values(['experiment', 'direction']).round(1).to_string(index=False))

calculate_improvements('shoreline_validation_results_expermient1-3.csv')
```
