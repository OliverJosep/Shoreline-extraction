## Load data

```python
patches = {
    "256x256": {
        "patch_size": (256, 256),
        "stride": (128, 128)
    },
    "256x512": {
        "patch_size": (256, 512),
        "stride": (128, 256)
    },
    "256x1024": {
        "patch_size": (256, 1024),
        "stride": (128, 512)
    }, 
    "512x512": {
        "patch_size": (512, 512),
        "stride": (256, 256)
    }
}

stations = ['agrelo', 'arenaldentem', 'cadiz', 'cies', 'samarador']
```

```python
for key in patches.keys():
    stations_count = {
        'agrelo': 0,
        'arenaldentem': 0,
        'cadiz': 0,
        'cies': 0,
        'samarador': 0
    }
    data_path = os.path.abspath(os.path.join(os.path.join(os.getcwd(), "../../.."), f"data/patchify_3_classes_rectified_{key}/"))
    print("Data path:", data_path)
    
    # get train/val/test splits
    files = os.listdir(data_path)
    for file in files:
        new_files = os.listdir(os.path.join(data_path, file, "images"))
        for new_file in new_files:
            for station in stations:
                if station in new_file:
                    stations_count[station] += 1
    print(f"Patch size: {key}")
    total = 0
    for station in stations:
        print(f"Station: {station}, count: {stations_count[station]}")
        total += stations_count[station]
    print(f"Total patches: {total}")

```
